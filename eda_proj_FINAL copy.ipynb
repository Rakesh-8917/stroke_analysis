#24. Outlier Analysis (IQR Method)
#***🎯 Objective***
Identify outliers in numerical features using the Interquartile Range (IQR) method.

```python
numerical_columns = ["age", "avg_glucose_level", "bmi"]

for column in numerical_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    outliers = df[(df[column] < lower) | (df[column] > upper)]

    print("=" * 50)
    print("Column :", column)
    print("Number of Outliers :", len(outliers))
```

#***📖 Observation***
The code reports the number of outliers in each numerical column.
#***💼 Business Insight***
Some extreme values may represent genuine high-risk patients rather than errors.

#1.1 Measures of Central Tendency
#***Objective***
Calculating the Mean, Median, and Mode of numerical features.


```python
# Measures of Central Tendency

numerical_columns = ['age', 'avg_glucose_level', 'bmi']

for column in numerical_columns:
    print("="*60)
    print("Column :", column)
    print("Mean   :", round(df[column].mean(),2))
    print("Median :", round(df[column].median(),2))
    print("Mode   :", df[column].mode()[0])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

numerical_columns = ['age', 'avg_glucose_level', 'bmi']

for column in numerical_columns:
    mean_val = df[column].mean()
    median_val = df[column].median()
    mode_val = df[column].mode()[0]

    # Create a DataFrame for plotting
    plot_data = pd.DataFrame({
        'Measure': ['Mean', 'Median', 'Mode'],
        'Value': [mean_val, median_val, mode_val]
    })

    plt.figure(figsize=(7, 5))
    sns.barplot(x='Measure', y='Value', data=plot_data, hue='Measure', palette='viridis', legend=False)
    plt.title(f'Measures of Central Tendency for {column.replace("_", " ").title()}')
    plt.ylabel('Value')
    plt.xlabel('Measure')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.show()
```

#***📖 Observation***
#Compare the mean and median of each numerical variable.
* If the mean is greater than the median, the data is likely right-skewed.
* If the mean is less than the median, the data is likely left-skewed.

#***💼 Business Insight***
Understanding the average patient characteristics helps hospitals identify normal health ranges and detect potential abnormalities.
#1.2 Measures of Dispersion
#***Objective***
Measure the spread of numerical variables.

```python
# Measures of Dispersion
import matplotlib.pyplot as plt
import seaborn as sns

# Prepare data for plotting
measures_data = {
    'Measure': [],
    'Column': [],
    'Value': []
}

# Assuming numerical_columns is defined in a previous cell (['age', 'avg_glucose_level', 'bmi'])
# If not, it should be re-defined here for self-containment.
numerical_columns = ['age', 'avg_glucose_level', 'bmi']

for column in numerical_columns:
    measures_data['Measure'].append('Variance')
    measures_data['Column'].append(column)
    measures_data['Value'].append(df[column].var())

    measures_data['Measure'].append('Standard Deviation')
    measures_data['Column'].append(column)
    measures_data['Value'].append(df[column].std())

    measures_data['Measure'].append('Minimum')
    measures_data['Column'].append(column)
    measures_data['Value'].append(df[column].min())

    measures_data['Measure'].append('Maximum')
    measures_data['Column'].append(column)
    measures_data['Value'].append(df[column].max())

measures_df = pd.DataFrame(measures_data)

# Create subplots for each numerical column
fig, axes = plt.subplots(nrows=len(numerical_columns), ncols=1, figsize=(10, 5 * len(numerical_columns)))
fig.suptitle('Measures of Dispersion for Numerical Columns', fontsize=16, y=1.02)

for i, column in enumerate(numerical_columns):
    subset_df = measures_df[measures_df['Column'] == column]
    sns.barplot(x='Measure', y='Value', data=subset_df, ax=axes[i], palette='viridis', hue='Measure', legend=False)
    axes[i].set_title(f'Column: {column}')
    axes[i].set_ylabel('Value')
    axes[i].set_xlabel('')
    axes[i].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
# Measures of Dispersion

for column in numerical_columns:
    print("="*60)
    print("Column :", column)
    print("Minimum :", df[column].min())
    print("Maximum :", df[column].max())
    print("Range   :", df[column].max()-df[column].min())
    print("Variance :", round(df[column].var(),2))
    print("Standard Deviation :", round(df[column].std(),2))
```

#***📖 Observation***
Large standard deviation indicates high variability.
Small standard deviation indicates data points are closer to the mean.
#***💼 Business Insight***

#1. Statistical Analysis

#1.1 Measures of Central Tendency


```python
# Measures of Central Tendency

numerical_columns = ['age', 'avg_glucose_level', 'bmi']

for column in numerical_columns:
    print("="*60)
    print("Column :", column)
    print("Mean :", round(df[column].mean(),2))
    print("Median :", round(df[column].median(),2))
    print("Mode :", df[column].mode()[0])                     ,
```

#***📖 Observation***
Compare mean, median, and mode.
If Mean > Median → Right Skewed.
If Mean < Median → Left Skewed.

Understanding average patient characteristics helps hospitals establish baseline health metrics and identify abnormal patterns.
#1.2 Measures of Dispersion

```python
for column in numerical_columns:

    print("="*60)
    print("Column :", column)

    print("Variance :", round(df[column].var(),2))
    print("Standard Deviation :", round(df[column].std(),2))
    print("Minimum :", df[column].min())
    print("Maximum :", df[column].max())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Prepare data for plotting
measures_data = {
    'Measure': [],
    'Column': [],
    'Value': []
}

for column in numerical_columns:
    measures_data['Measure'].append('Variance')
    measures_data['Column'].append(column)
    measures_data['Value'].append(df[column].var())

    measures_data['Measure'].append('Standard Deviation')
    measures_data['Column'].append(column)
    measures_data['Value'].append(df[column].std())

    measures_data['Measure'].append('Minimum')
    measures_data['Column'].append(column)
    measures_data['Value'].append(df[column].min())

    measures_data['Measure'].append('Maximum')
    measures_data['Column'].append(column)
    measures_data['Value'].append(df[column].max())

measures_df = pd.DataFrame(measures_data)

# Create subplots for each numerical column
fig, axes = plt.subplots(nrows=len(numerical_columns), ncols=1, figsize=(10, 5 * len(numerical_columns)))
fig.suptitle('Measures of Dispersion for Numerical Columns', fontsize=16, y=1.02)

for i, column in enumerate(numerical_columns):
    subset_df = measures_df[measures_df['Column'] == column]
    sns.barplot(x='Measure', y='Value', data=subset_df, ax=axes[i], palette='viridis')
    axes[i].set_title(f'Column: {column}')
    axes[i].set_ylabel('Value')
    axes[i].set_xlabel('')
    axes[i].tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

#***📖 Observation***
Variables with larger standard deviations indicate greater variability among patients.
#***💼 Business Interpretation***
Greater variability suggests that patient health conditions differ significantly, requiring personalized healthcare strategies.

#2. Hypothesis Testing

#2.1 Independent T-Test
Business Question
Do stroke and non-stroke patients have different average BMI values?
##Null Hypothesis (H₀)
There is no significant difference in BMI between stroke and non-stroke patients.
##Alternative Hypothesis (H₁)
There is a significant difference in BMI between stroke and non-stroke patients.

```python
stroke = df[df['stroke']==1]['bmi']

non_stroke = df[df['stroke']==0]['bmi']

t_stat, p_value = stats.ttest_ind(stroke,
                                  non_stroke,
                                  equal_var=False)

print("T Statistic :", t_stat)
print("P Value :", p_value)
```

##decission

```python
if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

#***💼 Business Interpretation***
A statistically significant result suggests BMI differs between stroke and non-stroke patients and should be considered in clinical assessments.
#2.2 Chi-Square Test
Business Question
Is hypertension associated with stroke?
##H₀
Hypertension and stroke are independent.
##H₁
Hypertension and stroke are associated.

```python
contingency = pd.crosstab(df['hypertension'],
                          df['stroke'])

chi2, p_value, dof, expected = stats.chi2_contingency(contingency)

if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")

print("Chi Square :", chi2)
print("P Value :", p_value)
```

#decission

```python
if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

#***💼 Business Interpretation***
If significant, hypertension is statistically associated with stroke and should be prioritized in preventive care.
#2.3 ANOVA Test
Business Question
Does average glucose level differ across work types?
##H₀
The mean glucose level is the same across all work types.
##H₁
At least one work type has a different mean glucose level.

```python
groups = []

for work in df['work_type'].unique():
    groups.append(
        df[df['work_type']==work]['avg_glucose_level']
    )

f_stat, p = stats.f_oneway(*groups)

print("F Statistic :", f_stat)
print("P Value :", p)
```

##decission

```python
if p < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")
```

#***💼 Business Interpretation***
A significant result suggests that average glucose levels differ across occupational groups.
#3. Confidence Interval
#***🎯 Objective***
Calculate the 95% confidence interval for the average BMI.
💻 Code


```python
confidence = 0.95

mean = df['bmi'].mean()

std = df['bmi'].std()

n = len(df)

margin = stats.norm.ppf(0.975) * (std/np.sqrt(n))

lower = mean - margin
upper = mean + margin

print("95% Confidence Interval")
print(lower, upper)
```

#***📖 Observation***
The confidence interval provides a range within which the true population mean BMI is expected to lie with 95% confidence.
